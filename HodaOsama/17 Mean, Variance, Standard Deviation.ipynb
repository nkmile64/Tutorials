<h3 style="color:#4fd7e7">What is the Mean of a Random Variable?</h3>

The **mean** (or expected value) of a random variable (x) is its **probability-weighted average** of all possible values.

<h4 style="color:#d89d49">1 | Mean of a Discrete Random Variable</h4>

<div style="text-align:center;">

$\mu_x= E(x) = \sum\limits_{i=1}^{n} x_i P(x_i)$

</div>
<img src="./img/discrete_random_var_mean_light.png" style=width:70vw;"">

<h4 style="color:#d89d49">2 | Mean of a Continuous Random Variable</h4>

$\mu_x = E(x) = \displaystyle\int_{-\infty}^{\infty} xf(x) \ dx$

Where, **f(x)** is the **Probability Density Function (PDF)**

<img src="./img/continuous_random_var_mean.png" style=width:70vw;>

<h4 style="color:#d89d49">3 | Mean Under Linear Transformations</h4>

If we transform X as:<br/>

$Y = a + bX$

then:

$E(Y) = a + bE(X)$

<img src="./img/linear_transformation_mean.png" style="width:70vw;">

<h3 style="color:#4fd7e7">What is Variance?</h3>
Variance measures the **spread** or **deviation** of values around the mean:
<h4 style="color:#d89d49">4 | Variance for a Discrete Random Variable</h4>

$Var(x) = \sum\limits_{i}(x_i - \mu)^2 P(x_i)$

<img src="./img/discrete_variance_contributions.png" style="width:70vw;">

<h4 style="color:#d89d49">5 | Variance for a Continuous Random Variable</h4>

$Var(x) = \displaystyle\int_{-\infty}^{\infty} (x - \mu)^2 f(x) \ dx$

<h4 style="color:#d89d49">6 | Variance Under Linear Transformations</h4>

For, Y = a + bX, variance changes as:

$Var(Y) = b^2Var(X)$

<h3 style="color:#4fd7e7">SUMMARY</h3>

- Adding a constant a to a random variable, shifts the mean but leaves variance *unchanged*.
- Multiplying by b, scales the variance by $b^2$
- Standard Deviations scales by the absolute value of b: $|b|$
