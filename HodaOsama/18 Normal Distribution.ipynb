**Normal distribution** or **Gaussian distribution** is a type of continuous probability distribution that is *symmetric* and *bell-shaped*.

The equation for the PDF (Probability Density Function) of a Normal Distribution is:
<div style="text-align:center;">

$f(x) = \Large \frac{1}{\sigma \sqrt{2 \pi}} e^{-(\frac{(x - \mu)^2}{2 \sigma^2})}$

</div>

- $\mu$ is the **mean** of the distribution: where the center of the bell curve lies.
- $\sigma$ is the **standard deviation**: how spread out the values are around the mean.
- $e$ is the **exponential function**

The **area** under the curve represents the total probability that equals to 1.<br/>
Since x can take any value, the distribution extends infinitely in both directions.

#### Visualizing the 68-95-99.7 Rule

For any normal distribution:
- **68%** of values lies between $\mu - \sigma \ and \ \mu + \sigma$
- **95%** of values lies between $\mu - 2\sigma \ and \ \mu + 2\sigma$
- **99.7%** of values lies between $\mu - 3\sigma \ and \ \mu + 3\sigma$
