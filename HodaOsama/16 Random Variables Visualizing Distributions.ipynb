# What is a Random Variable?

A **random variable** is a numerical outcome of a random phenomenon.

- Table

# Discrete Probabilities are Simpler

With discrete random variables we just **add up the values:**<br/>

$P(X=2 \ or \ X=3) = P(X=2) + P(X=3)$

# CDF - Cumulative Distribution Function

> What is the probability that X is less than or equal to some value?

**Example: CDF (Discrete)**<br/>

Each step adds the probability from the previous value.<br/>
For example, the probability of an outcome being 4, is P(X=4) = 0.25 by itself.<br/>
However, the cumulative probability of an outcome being 4 or less, is 0.85.<br/>
